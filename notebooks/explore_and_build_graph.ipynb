```python
from scripts.data_extraction import JiraMongoExtractor
from scripts.graph_construction import GraphBuilder, attach_embeddings_to_graph
from scripts.text_embedding import TextEmbedder
```

```python
extractor = JiraMongoExtractor(db_name="JiraRepos")
issues = extractor.fetch_issues("Jira", projection={"fields.summary":1,"fields.description":1,"fields.issuelinks":1,"id":1,"key":1})
extractor.close()

# Build
builder = GraphBuilder()
G = builder.build_graph(issues)
len(G.nodes()), len(G.edges())

# Embeddings
embedder = TextEmbedder()
attach_embeddings_to_graph(G, embedder)

# Just check a random node's embedding
random_node = next(iter(G.nodes()))
print(G.nodes[random_node]["embedding"])
```
